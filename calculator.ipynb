```python
n = 1
while n == 1:
    try:
        # Taking first number from user
        num1 = input("Enter the first number: ")
        #convert string into float
        num1 = float(num1)
    except ValueError:
        print("Enter a valid number!\n")
        continue

    # Menu of operators
    print("\nOperators:\n addition (+)\n subtraction (-)\n multiplication (*)\n division (/)\n floor division (//)\n modulus (%)\n exponentiation (**)\n")

    # Taking operator from user
    operator = input("Enter the operator: ").strip()

    try:
        # Taking second number from user
        num2 = input("Enter the second number: ")
        #convert string into float
        num2 = float(num2)
    except ValueError:
        print("Enter a valid number!\n")
        continue

    # Calculation and validation of operator
    if operator == '+':
        result = num1 + num2
    elif operator == '-':
        result = num1 - num2
    elif operator == '*':
        result = num1 * num2
    elif operator == '/':
        if num2 == 0:
            print("Error: Cannot divide by zero!\n")
            continue
        result = num1 / num2
    elif operator == '//':
        if num2 == 0:
            print("Error: Cannot integer-divide by zero!\n")
            continue
        result = num1 // num2
    elif operator == '%':
        if num2 == 0:
            print("Error: Cannot calculate modulo by zero!\n")
            continue
        result = num1 % num2
    elif operator == '**':
        result = num1 ** num2
    else:
        print("Invalid operator!\n")
        continue

    # display the output clearly using formatted print statements
    print(f"\nResult: {num1} {operator} {num2} = {result}\n")

    # Exit condition
    again = input("Do you want to calculate again? (Y/N): ").strip()
    if again != 'Y' and again != 'y':
        print("Have a nice day!")
        break
```
